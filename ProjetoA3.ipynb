```python
# Bloco 14
# Modelagem - iniciando a Arvore de Decisão
# Classificação do PIB com base no FMI (Fundo Monetário Internacional)
# 0 - PIB Baixo até US$ 100 bilhões
# 1 - PIB Médio US$ 100 bilhões até US$ 1 trilhão
# 2 - PIB Alto acima de US$ 1 trilhão

def classificar_pib(valor):

    if valor < 100_000_000_000:
        return 0   # PIB Baixo

    elif valor < 1_000_000_000_000:
        return 1   # PIB Médio

    else:
        return 2   # PIB Alto

```

```python
# Bloco 15
# Aplicando ao data Frame
GlobEI['Classe_Economia'] = GlobEI[
    'Gross Domestic Product (GDP)'
].apply(classificar_pib)
```

```python
# Bloco 16
# X -Atributo e Y - variavel alvo (target)
X = GlobEI[['Gross Domestic Product (GDP)']]
y = GlobEI['Classe_Economia']
```

```python
# Bloco 17
# Definindo proporção usadas para teste 25% e 75% de treino.
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25)
```

```python
# Bloco 18
# Previsões usando o método predict
from sklearn import tree
clf = tree.DecisionTreeClassifier(criterion='entropy', max_depth=3)
clf = clf.fit(X_train, y_train)
predictions = clf.predict(X_test)
```

```python
# Bloco 19
# Matriz de Confusão
print("\nMatriz de confusão detalhada:\n", pd.crosstab(y_test, predictions, rownames=['Real'], colnames=['Predito'], margins=True, margins_name='Todos'))
```

```python
# Bloco 20
# Medidas usadas em avaliação:
import sklearn.metrics as metrics
print("Relatório sobre a Classificação:\n")
print(metrics.classification_report(y_test, predictions, target_names=['PIB Baixo', 'PIB Médio', 'PIB Alto']))
```

```python
# Bloco 21
# Árvore para o formato Graphviz
import graphviz
dot_data = tree.export_graphviz(clf, out_file=None)
graph = graphviz.Source(dot_data)
graph.render("Class")
dot_data = tree.export_graphviz(clf, out_file=None, feature_names=X.columns, class_names=['PIB Baixo', 'PIB Médio', 'PIB Alto'], filled=True, rounded=True, special_characters=True)
graph = graphviz.Source(dot_data, format="png")
graph
```

```python
# Bloco 36
# APLICANDO O ALGORITMO DA ÁRVORE DE DECISÃO #
from sklearn import tree
modeloARV = tree.DecisionTreeClassifier(criterion='entropy', max_depth=3)
modeloARV = modeloARV.fit(X_train, y_train)
predictionsARV = modeloARV.predict(X_test)

# MOSTRANDO OS RESULTADOS DO ALGORITMO DA ÁRVORE DE DECISÃO #

print("\nMatriz de confusão detalhada da Árvore de Decisão:\n",
      pd.crosstab(y_test, predictionsARV, rownames=['Real'], colnames=['Predito'],
      margins=True, margins_name='Todos'))

dic_arv = metrics.classification_report(y_test, predictionsARV, target_names=['PIB Baixo', 'PIB Médio', 'PIB Alto'],output_dict=True)
result_arv = dic_arv['accuracy']
print("Acurácia da Árvore de Decisão: ", result_arv)

```

```python
# Bloco 37
# APLICANDO O ALGORITMO KNN #

from sklearn.neighbors import KNeighborsClassifier

modeloKNN = KNeighborsClassifier(n_neighbors=3)
modeloKNN.fit(X_train, y_train)
predictionsKNN = modeloKNN.predict(X_test)

# MOSTRANDO OS RESULTADOS DO ALGORITMO KNN #

print("\nMatriz de confusão detalhada do KNN:\n",
      pd.crosstab(y_test, predictionsKNN, rownames=['Real'], colnames=['Predito'],
      margins=True, margins_name='Todos'))

dic_knn = metrics.classification_report(y_test, predictionsKNN, target_names=['PIB Baixo', 'PIB Médio', 'PIB Alto'],output_dict=True)
result_knn = dic_knn['accuracy']
print("Acurácia do KNN: ", result_knn)
```

```python
# Bloco 38
# APLICANDO O ALGORITMO NAIVES #

from sklearn.naive_bayes import GaussianNB

modeloNAIVES = GaussianNB()
modeloNAIVES.fit(X_train, y_train)
predictionsNAIVES = modeloNAIVES.predict(X_test)

# MOSTRANDO OS RESULTADOS DO ALGORITMO NAIVES #

print("\nMatriz de confusão detalhada do NAIVES:\n",
      pd.crosstab(y_test, predictionsNAIVES, rownames=['Real'], colnames=['Predito'],
      margins=True, margins_name='Todos'))

dic_NAIVES = metrics.classification_report(y_test, predictionsNAIVES, target_names=['PIB Baixo', 'PIB Médio', 'PIB Alto'],output_dict=True)
result_NAIVES = dic_NAIVES['accuracy']
print("Acurácia do NAIVES: ", result_NAIVES)
```

```python
# Bloco 39
# COMITÊ DE CLASSIFICADORES

print("COMITÊ DE CLASSIFICADORES")

# Exemplo de PIB para classificação com PIB de 500 bilhões.
novo_pib = pd.DataFrame({
    'Gross Domestic Product (GDP)': [500000000000]
})

tot_baixo = 0
tot_medio = 0
tot_alto = 0

# ÁRVORE DE DECISÃO
pred_arv = modeloARV.predict(novo_pib)[0]

if pred_arv == 0:
    tot_baixo += 1
elif pred_arv == 1:
    tot_medio += 1
else:
    tot_alto += 1

print("Árvore de Decisão:", pred_arv)

# KNN
pred_knn = modeloKNN.predict(novo_pib)[0]

if pred_knn == 0:
    tot_baixo += 1
elif pred_knn == 1:
    tot_medio += 1
else:
    tot_alto += 1

print("KNN:", pred_knn)

# NAIVE BAYES
pred_nb = modeloNAIVES.predict(novo_pib)[0]

if pred_nb == 0:
    tot_baixo += 1
elif pred_nb == 1:
    tot_medio += 1
else:
    tot_alto += 1

print("Naive Bayes:", pred_nb)

# RESULTADO FINAL

print("\nTotal de votos:")
print("PIB Baixo =", tot_baixo)
print("PIB Médio =", tot_medio)
print("PIB Alto  =", tot_alto)

if tot_baixo >= 2:
    print("\nO comitê acredita que a economia possui PIB BAIXO.")

elif tot_medio >= 2:
    print("\nO comitê acredita que a economia possui PIB MÉDIO.")

else:
    print("\nO comitê acredita que a economia possui PIB ALTO.")

# Verifica unanimidade

if tot_baixo == 3:
    print("O comitê foi unânime: PIB BAIXO.")

elif tot_medio == 3:
    print("O comitê foi unânime: PIB MÉDIO.")

elif tot_alto == 3:
    print("O comitê foi unânime: PIB ALTO.")
```
